```
# A. Load the Iris dataset

from sklearn.datasets import load_iris

iris = load_iris()

print(iris)
```

```
print("Feature names:", iris.feature_names)
print("Target names:", iris.target_names)
print("Data shape:", iris.data.shape)
print("Target shape:", iris.target.shape)
```

```
# B. Set X to feature matrix

X = iris.data

print(X.shape)
```

```
# C. Set y to target vector

y = iris.target

print(y)
```

```
# D. Scale the data (StandardScaler)

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print(X_scaled[:5])
```

```
# E. Fit DBSCAN and get resulting groups

from sklearn.cluster import DBSCAN

dbscan = DBSCAN(
    eps=0.5,
    min_samples=5
)

clusters = dbscan.fit_predict(X_scaled)

print(clusters)
```

```
import numpy as np

n_clusters = len(set(clusters)) - (1 if -1 in clusters else 0)

print("Number of clusters:", n_clusters)
```

```
n_noise = np.sum(clusters == -1)

print("Number of noise points:", n_noise)
```

```
# F. Find the most occurring flower type for each group

from scipy.stats import mode
import numpy as np

unique_clusters = set(clusters)

mapped_clusters = np.full_like(clusters, -1)

for cluster in unique_clusters:

    # Ignore noise
    if cluster == -1:
        continue

    mask = clusters == cluster

    most_common = mode(
        y[mask],
        keepdims=True
    ).mode[0]

    mapped_clusters[mask] = most_common

print(mapped_clusters)
```

```
# G. Confusion Matrix and Accuracy
# Confusion Matrix

from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

# Remove noise for this evaluation
mask = clusters != -1

cm = confusion_matrix(
    y[mask],
    mapped_clusters[mask]
)

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=iris.target_names,
    yticklabels=iris.target_names
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("DBSCAN Confusion Matrix")

plt.show()
```

```
# G. Confusion Matrix and Accuracy
# Accuracy

from sklearn.metrics import accuracy_score

accuracy = accuracy_score(
    y[mask],
    mapped_clusters[mask]
)

print("DBSCAN Accuracy:", accuracy)
print("DBSCAN Accuracy (%):", accuracy * 100)
```

```
# H. Clustering Evaluation Metrics

# Metrics WITH true y

# 1. Adjusted Rand Index

from sklearn.metrics import adjusted_rand_score

ari = adjusted_rand_score(
    y[mask],
    clusters[mask]
)

print("Adjusted Rand Score:", ari)
```

```
# H. Clustering Evaluation Metrics

# Metrics WITH true y

# 2. V-measure

from sklearn.metrics import v_measure_score

v_measure = v_measure_score(
    y[mask],
    clusters[mask]
)

print("V-measure:", v_measure)
```

```
# H. Clustering Evaluation Metrics

# Metrics WITHOUT true y

# 3. Silhouette Score

from sklearn.metrics import silhouette_score

silhouette = silhouette_score(
    X_scaled[mask],
    clusters[mask]
)

print("Silhouette Score:", silhouette)
```

```
# J. Find a good eps

from sklearn.neighbors import NearestNeighbors
import numpy as np
import matplotlib.pyplot as plt

neighbors = NearestNeighbors(n_neighbors=5)

neighbors_fit = neighbors.fit(X_scaled)

distances, indices = neighbors_fit.kneighbors(X_scaled)

distances = np.sort(distances[:, 4])

plt.figure(figsize=(8, 5))

plt.plot(distances)

plt.xlabel("Data Points")
plt.ylabel("5th Nearest Neighbor Distance")
plt.title("DBSCAN k-Distance Plot")

plt.grid(True)

plt.show()
```

```
# K. Try different eps

eps_values = [0.3, 0.4, 0.5, 0.6, 0.7, 0.8, 1.0]

for eps in eps_values:

    db = DBSCAN(
        eps=eps,
        min_samples=5
    )

    labels = db.fit_predict(X_scaled)

    n_clusters = len(
        set(labels)
    ) - (1 if -1 in labels else 0)

    n_noise = np.sum(labels == -1)

    print(
        "eps =", eps,
        "| clusters =", n_clusters,
        "| noise =", n_noise
    )
```

```
# L. Silhouette Score for DBSCAN

eps_values = np.arange(0.2, 1.1, 0.1)

scores = []

for eps in eps_values:

    db = DBSCAN(
        eps=eps,
        min_samples=5
    )

    labels = db.fit_predict(X_scaled)

    mask = labels != -1

    # Need at least 2 clusters
    if len(set(labels[mask])) >= 2:

        score = silhouette_score(
            X_scaled[mask],
            labels[mask]
        )

        scores.append(score)

    else:
        scores.append(np.nan)
```

```
# L. Silhouette Score for DBSCAN (Plot)

plt.figure(figsize=(8, 5))

plt.plot(
    eps_values,
    scores,
    marker='o'
)

plt.xlabel("eps")
plt.ylabel("Silhouette Score")
plt.title("DBSCAN: Silhouette Score vs eps")

plt.grid(True)

plt.show()
```

```
# M. DBSCAN without scaling

db_original = DBSCAN(
    eps=0.5,
    min_samples=5
)

clusters_original = db_original.fit_predict(X)
```

```
# N. DBSCAN with scaling

db_scaled = DBSCAN(
    eps=0.5,
    min_samples=5
)

clusters_scaled = db_scaled.fit_predict(X_scaled)
```

```
# O. Important Metrics

from sklearn.cluster import DBSCAN
from sklearn.metrics import (
    adjusted_rand_score,
    v_measure_score,
    silhouette_score,
    accuracy_score
)
from scipy.stats import mode
import numpy as np

eps_values = [0.3, 0.4, 0.5, 0.6, 0.7, 0.8, 1.0]

for eps in eps_values:

    dbscan = DBSCAN(
        eps=eps,
        min_samples=5
    )

    labels = dbscan.fit_predict(X_scaled)

    # Number of clusters
    n_clusters = len(set(labels)) - (1 if -1 in labels else 0)

    # Noise
    n_noise = np.sum(labels == -1)

    # Remove noise for evaluation
    mask = labels != -1

    # Need at least 2 clusters
    if n_clusters >= 2 and np.sum(mask) > 0:

        # Map clusters to most common actual flower
        mapped = np.full_like(labels, -1)

        for cluster in set(labels):
            if cluster == -1:
                continue

            cluster_mask = labels == cluster

            mapped[cluster_mask] = mode(
                y[cluster_mask],
                keepdims=True
            ).mode[0]

        accuracy = accuracy_score(
            y[mask],
            mapped[mask]
        )

        ari = adjusted_rand_score(
            y[mask],
            labels[mask]
        )

        v_measure = v_measure_score(
            y[mask],
            labels[mask]
        )

        silhouette = silhouette_score(
            X_scaled[mask],
            labels[mask]
        )

    else:
        accuracy = np.nan
        ari = np.nan
        v_measure = np.nan
        silhouette = np.nan

    print(
        f"eps={eps:.1f} | "
        f"clusters={n_clusters} | "
        f"noise={n_noise} | "
        f"accuracy={accuracy:.3f} | "
        f"ARI={ari:.3f} | "
        f"V-measure={v_measure:.3f} | "
        f"silhouette={silhouette:.3f}"
    )
```

```
# P. Plot DBSCAN clusters for ALL eps

import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN

eps_values = [0.3, 0.4, 0.5, 0.6, 0.7, 0.8, 1.0]

plt.figure(figsize=(15, 10))

for i, eps in enumerate(eps_values):

    dbscan = DBSCAN(
        eps=eps,
        min_samples=5
    )

    clusters = dbscan.fit_predict(X_scaled)

    plt.subplot(3, 3, i + 1)

    plt.scatter(
        X_scaled[:, 2],       # Petal Length
        X_scaled[:, 3],       # Petal Width
        c=clusters,
        cmap='viridis',
        s=40
    )

    plt.xlabel("Petal Length (scaled)")
    plt.ylabel("Petal Width (scaled)")
    plt.title(f"DBSCAN: eps = {eps}")

plt.tight_layout()
plt.show()
```
