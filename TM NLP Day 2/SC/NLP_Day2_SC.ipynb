# Text Cleaning & Preprocessing

Since text is the most unstructured form of all the available data, various types of noise are present in it and the data is not readily analyzable without any pre-processing. The entire process of cleaning and standardization of text, making it noise-free and ready for analysis is known as text preprocessing.

It is comprised of two steps:

- <b>Noise Removal</b>
    - Language stopwords (commonly used words of a language – is, am, the, of, in etc) 
    - Punctuations
    - URLs or links
    - Social media entities (mentions, hashtags)
    - Tags
    - Numbers
    - Common & rare words
    - Long & short words
    - non-ASCII Characters
    - Spelling correction
<br> <br>    
- <b>Lexicon Normalization</b>
     - Stemming
     - Lemmatization
     
All these pre-processing steps are essential and help us in reducing our vocabulary clutter so that the features produced in the end are more effective.

## Lower Case

For some NLP tasks, before starting pre-processing stage, we will transform our text into lower case. This avoids having multiple copies of the same words. For example, while calculating the word count of a text, ‘Process’ and ‘process’ will be taken as different words.

# Noise Removal

Any piece of text which is not relevant to the context of the data and the end-output can be specified as the noise.

This step deals with removal of all types of noisy entities present in the text.

A general approach for noise removal is to prepare a dictionary of noisy entities, and iterate the text object by tokens (or by words), eliminating those tokens which are present in the noise dictionary.

## Filter Out Punctuation
The next step is to remove punctuation, as it does not add any extra information while treating text data. Therefore, removing all instances of it will help us reduce the size of data. Some examples of punctuation are like commas, apostrophes, quotes, question marks, and more. `string.punctuation` provides a great list of punctuation characters.

````python
import string
print(string.punctuation)
````

Python has the function `isalpha()` that can be used. The `isalpha()` method returns True if all the characters are alphabet letters (a-z).

````python
# Remove all punctuations 
text = "This &is [an] example? {of} string. with?. punctuation!!!! won't"

text_tokens = word_tokenize(text)
print(text_tokens)
[w for w in text_tokens if w.isalpha()] 
# isalpha() checks whether the string consists of alphabetic characters only
````

While using the function `isalpha()` is able to remove all punctuations from the text, but there is one issue with this method. The word ”won't” **is removed** from the text. This will change the sentiment of the sentence!

Hence using the `isalpha()` method is only suitable for NLP tasks such as content clustering or topic modelling. For sentiment analysis related tasks, we should use methods that will ignore such punctuation. There are two ways to solve this issue:
  1. We can use the `sub()` method of regex to remove only punctuation characters, instead of removing words that contain a punctuation.
  2. We can use the `translate()` method of the `string` module. For more detail, you can check the [Extra Study part](#Filter-Out-Punctuation-using-translate-method).

## Common & Rare Words Removal

The main question is why do we want to remove frequent and infrequent words?

This is very intuitive, as some of the words that are very unique in nature like names, brands, and product names need to be removed for different NLP tasks. Because they are so rare, the association between them and the other words is dominated by noise. For example, it would be really bad to use names as a predictor for a text classification problem, even if they come out as a significant predictor. We definitely do not want all these noisy tokens to be present.

Similarly as rare words, many articles have mentioned to remove the most common words from the text because the most frequent words are almost devoided of meaning and their presence will not be of any use in classification of the text. In addition, they really hurt computational speed. If we could reduce the size of out data, just by removing frequent words, it could drastically improve performance of our model.


### Remove Most Frequent Words in Corpus

We have already learnt how to use frequency distribution to record the number of times each word has occurred in a text. The frequency distribution can then be used to remove the most frequent words in a corpus. We can get the most frequently occuring words in the corpus from the distribution and create a vocabulary containing the most frequent words. Then we remove the most frequent words from the corpus similarly as how we remove the previous noises.

````python
from nltk.probability import FreqDist

text = "I am happy. You are sad. I am not like you. You are crazy. I am not crazy. I do not like you."   
token = text.split(' ')

freq_dist = FreqDist(token)
print(list(freq_dist))
````

````python
# The 3 most frequently occurring words in our text data
freq_most_common = freq_dist.most_common(3)
print(freq_most_common)
````

````python
# Create a vocabulary from 3 most frequent words
common_words = [w for w,freq in freq_most_common]
print(common_words)
````

````python
# Remove frequent words
after_common_words = " ".join([ word for word in token if word not in common_words])
after_common_words
````

### Remove Rare Words from the Corpus

We can remove rare words from the corpus similarly to how we remove frequent words from the corpus, using frequency distribution. Instead of obtaining the most frequent words in the text, we get the count of each unique word in the text. We create a vocabulary of rare words based on the count of each words. Then we remove all the words in the vocabulary from the text.

````python
# Remove rare words in corpus
text = "I am happy You are sad I am not like you You are crazy I am not crazy I do not like you"   
tokens = word_tokenize(text)

freq_dist = FreqDist(tokens)
print(list(freq_dist))
````

````python
freq = list(freq_dist.values())
print(freq)
print()

keys = list(freq_dist.keys())
print(keys)
````

````python
# Create a vocabulary from rare words (one or two occurrence)
rare_words = [keys[i] for i, val in enumerate(freq) if val in range(1,3)]
print(rare_words)
````

````python
# Remove infrequent words from the text
after_rare_words = " ".join([ word for word in tokens if word not in rare_words])
after_rare_words
````

## Lexicon Normalization

Another type of textual noise is about the multiple representations exhibited by single word.

For example – “play”, “player”, “played”, “plays” and “playing” are the different variations of the word – “play”, even though they have different meaning but contextually all are similar. We want to convert all the disparities of a word into their normalized form (also known as lemma).

The most common lexicon normalization practices are :

- <b>Stemming</b>   
    
- <b>Lemmatization</b>

### Stemming
Stemming is the process of cutting down the words to their word stem, base or root form (for example “fishing”, “fished” and “fisher” all reduce to the stem “fish”). 

The main algorithm is the [Porter stemming algorithm](https://tartarus.org/martin/PorterStemmer/) which removes common morphological and inflexional endings (“ing”, “ly”, “es”, “s” etc) from words. Stemmer is easier to implement and the processing time is faster compared to lemmatizer. So stemming is used most of the time for its simplicity.

To use the Porter stemming algorithm, we import the `PorterStemmer()` class from the nltk module. We can then perform stemming using the `stem()` method provided by a `PorterStemmer()` object.

````python
from nltk.stem import PorterStemmer

new_text = "the little yellow dog barked at the cat. thinking, plays, cats, leaves"
# new_text = "There are several types of stemming algorithms."

words = word_tokenize(new_text)
ps = PorterStemmer()

words_stem = [ps.stem(w) for w in words]
print(words_stem)
````

````python
example_words = ["python","pythoner","pythoning","pythoned",
                 "pythonly", "pythons", 'eventuellement', 'darfed']

example_words_stem = [ps.stem(w) for w in example_words]
print(example_words_stem)
````

### Lemmatization
The aim of lemmatization, like stemming, is to reduce inflectional forms to a common base form. However, lemmatization is a more effective option compared to stemming because it **converts the word into its root word**, rather than just stripping the suffices. Lemmatization uses lexical knowledge bases to get the correct base forms of words. It makes use of the vocabulary (dictionary importance of words) and does a morphological analysis (word structure and grammar relations) to obtain the root word. Therefore, we usually prefer using lemmatization over stemming.

Lemmatization uses context and part of speech to determine the inflected form of the word and applies different normalization rules for each part of speech to get the root word (lemma). To use lemmatization, we can import the `WordNetLemmatizer()` class from the nltk module. We can then perform lemetization using the `lemmatize()` method provided by a `WordNetLemmatizer()` object.

````python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
words =("cats","geese","rocks","lives","leaves","worse","runing")
# words = ["been", "had", "done", "languages", "cities", "mice"]

words_lem = [lemmatizer.lemmatize(w) for w in words]
print(words_lem)
````

````python
print(ps.stem('cities'))
print(lemmatizer.lemmatize('cities'))
````

The `lemmatize()` method can take as a parameter the part of speech of the root word that we want to obtain. The part of speech constants to be enter as parameters are as follow:

- 'a': ADJECTIVE
- 'r': ADVERB
- 'n': NOUN
- 'v': VERB

Note: We will learn more about part of speech in the next section.

````python
print(lemmatizer.lemmatize("better", pos="a"))
print(lemmatizer.lemmatize("best", pos="a"))
print(lemmatizer.lemmatize("worse", pos="a"))
print(lemmatizer.lemmatize("runing"))
print(lemmatizer.lemmatize("runing", pos='v'))
````

#  Phrase Detection

Phrase detection is the process of finding phrases in general from a corpus. Queries made to search engines are usually longer than a single word in length. Using phrase detection will help in such task by reducing the average response time of the search engine.

Noun phrase detection is a topic that has been studied a lot in the NLP community.

One way for Phrase Detection is used to apply 'Chunking' and 'Chinking'.

## Chunking 
Chunking is a natural language process that identifies constituent parts of sentences (nouns, verbs, adjectives, etc.) and links them to higher order units that have discrete grammatical meanings (noun groups or phrases, verb groups, etc. In other words, chunk is a process of **meaningful extraction of short phrases from the sentence** (tagged with Part-of-Speech). Chunks are made up of words (group words) and the kinds of words are defined using the part-of-speech tags.

One of the main goals of chunking is to group words into what are known as "noun phrases." These are phrases of one or more words that contain:
  - a noun 
  - maybe some descriptive words
  - maybe a verb 
  - maybe something like an adverb 
        
The rules that make up a chunk grammar use **chunk patterns** to describe sequences of tagged words. 

### Chunk Patterns
Chuck patterns are normal regular expressions which are modified and designed to match the part-of-speech tag designed to match sequences of part-of-speech tags. Angle brackets are used to specify an indiviual tag. One can define multiple tags in the same way.

In order to chunk, we combine the **part of speech tags** with **regular expressions**.
An example of chunk pattern,
```python
                    <DT>?<NN.*>+
```
Curly Braces {} are used to specify a chunk. Now we have a chunk,
```python
                    {<DT>?<NN.*>+}
```

The `RegexpParser` is class is a grammar based chunk parse. In order to create a chunker, we first define a **chunk grammar** consisting of the rules that indicate how the text should be chunked. We then give the `RegexpParser` a the chunk grammar as an input to build a parser object which we can then chunk any given text using the `parse()` method.

````python
sample_text = "the little yellow dog barked at the cat"
# sample_text = "A black television and a white stove were bought for the new apartment of John."
words = word_tokenize(sample_text)

# The first step is to determine the part of speech for each word
tagged = pos_tag(words)
````

````python
# The second step is chunking
chunkGram = "Chunk: {<DT>?<JJ>*<NN>}"   # a chunk
chunkParser = nltk.RegexpParser(chunkGram)
chunked = chunkParser.parse(tagged)
print(chunked)
````

Given that a chunked text result in a tree, we can use the `draw()` method to display it graphically.

````python
chunked.draw()
````

Let us try another example which has two chunk patterns in a chunk.

````python
sample_text="Rapunzel let down her long golden hair"
words = word_tokenize(sample_text)

tagged = nltk.pos_tag(words)
    
chunkGram = r"""
  NP: {<DT>?<JJ>*<NN>}   
      {<NNP>}"""

chunkParser = nltk.RegexpParser(chunkGram)
chunked = chunkParser.parse(tagged)
print(chunked)
````

## Chinking
One can define a pattern or words that can’t be a part of chuck and such words are known as chinks. The flipped braces `}{` are used to specify the chink pattern. 

For a particular phrase type chunk and a chink pattern can be combined into a grammar. Chinking is performed after chunking. 

````python
sample_text="the little yellow dog barked at the cat"
words = word_tokenize(sample_text)

tagged = nltk.pos_tag(words)
chunkGram = r"""Chunk: {<DT>?<JJ>*<NN>}   
                        }<DT>{ """        # a chink
chunkParser = nltk.RegexpParser(chunkGram)
chunked = chunkParser.parse(tagged)
print(chunked)
chunked.draw()
````

Let us have an another example for chunking and chinking.

````python
sample_text="the little yellow dog barked at the cat"
words = word_tokenize(sample_text)

tagged = nltk.pos_tag(words)
chunkGram = r"""Chunk: {<.*>+}
                        }<VB.?|IN|DT|TO>+{ """
chunkParser = nltk.RegexpParser(chunkGram)
chunked = chunkParser.parse(tagged)
print(chunked)
chunked.draw()
````

## Extract the Chunks

In some cases, we are interested to extract the chunks from the text. For example, we know that the objects that we are interested in a text is defined in the chunk grammar, and would like to obtain the object as well as anything that describes the object (its associated adjectives).

````python
ch = []
for chunk in chunked:       # for each item in chunked
    if hasattr(chunk, 'label'):  # does chunk object have label attribute
        word = ' '.join(c[0] for c in chunk)    # try c[1]   # tuple structure
        ch.append(word)
        word = ""
ch
````

# Named Entity Recognition
In any text document, there are particular terms that represent specific entities that are more informative and have a unique context. These entities are known as **named entities**, which more specifically refer to terms that represent real-world objects like people, places, organizations, and so on. A naive approach could be to find these by looking at the noun phrases in text documents.

Named entity recognition (NER) is a popular technique used in information extraction to identify and segment the named entities and classify or categorize them into pre-defined categories (names of persons, locations, organizations, times, etc).

We can use the `ne_chunk` function of NLTK to extract named entity recognition out of a text. The function uses NLTK's currently recommended named entity chunker to chunk the given list of tagged tokens. The following are examples of named entity types in NLTK:
- Person names - <font color=green> Eddy Bonte, President Obama</font>
- Location names - <font color=green>Murray River, Mount Everest</font>
- Company names -<font color=green> Washington Monument, Stonehenge</font>
- Date - <font color=green>June, 2008-06-29</font>
- Time - <font color=green>two fifty a m, 1:30 p.m.</font>
- Percent - <font color=green>twenty pct, 18.75 %</font>
- Money - <font color=green>175 million Canadian Dollars, GBP 10.40</font>
- GPE (Geographical Entity) - <font color=green>South East Asia, Midlothian</font>
- Organization -<font color=green> Georgia-Pacific Corp., WHO</font>

### Displaying emoji in Python

The entire set of emoji codes as defined by the [unicode consortium](http://www.unicode.org/emoji/charts/full-emoji-list.html), and is supported in addition to a bunch of [aliases](https://www.webfx.com/tools/emoji-cheat-sheet/). We can use the `emojize()` function to turn the *unicode consortiums* or the *aliases* into their corresponding emojis. By default only the official list is enabled but inserting the parameter `use_aliases=True` to the function enables both the full list and aliases.


````python
import emoji

# Display emojis using their unicodes
print(emoji.emojize("Printing out emojis in Python is easy! \U0001f604 \U0001f4af \U0001f44f"))

# Display emoji using their aliases
print(emoji.emojize("Printing out emojis in Python is easy! :thumbsup: :laughing: :office:", use_aliases=True))
````

### Extracting emojis

In some NLP tasks, we want to extract out emojis as they might be useful for sentiment analysis such as the emotions the text is trying to evoke. We can do this by checking whether each character in the text is an emoji using the `emoji.UNICODE_EMOJI` dictionary.


````python
text = "This 🤔sentence 🙈has 😌a 💕lot 🙅of 👩emoji 😊"
print(text)
````

````python
# Extract the emojis
emojis = ''.join(char for char in text if char in emoji.UNICODE_EMOJI)
print(emojis)
````

### Removing emojis

In some other cases, emojis do not provide any useful information for the NLP tasks. In this case, we treat emojis as noise and remove them from the text

````python
# Remove the emojis
removed = ''.join(char for char in text if char not in emoji.UNICODE_EMOJI)
print(removed)
````

## Spelling Correction with pyspellchecker

Another way for spell checking in Python is `SymSpell` (based on the Symmetric Delete spelling correction algorithm) method that is **fast**. This library is based on Peter Norvig's implementation. This method uses a [Levensthein Distance](https://en.wikipedia.org/wiki/Levenshtein_distance) algorithm to find permutations within an edit distance of 2 from the original word. It then compares all permutations to known words in a word frequency list. Those words that are found more often in the frequency list are more likely the correct results. 

To use this method, we need to install the `pyspellchecker` package firstly. After installation, using `pyspellchecker` should be fairly straight forward. We just need to import the `SpellChecker()` class from the `spellchecker` module. The `SpellChecker()` object returns the default word frequency list.

Below contains some of the available methods:

- <font color=green><b>spell.known([words])</b></font>:
This method returns those words that are in the word frequency list

- <font color=green><b>spell.unknown([words])</b></font>:
This method returns those words that are not in the frequency list

- <font color=green><b>spell.correction(word)</b></font>:
This method returns the most probable result for the misspelled word

- <font color=green><b>spell.candidates(word)</b></font>:
This method returns a set of possible candidates for the misspelled word


````python
# Install pyspellchecker package
# !pip install pyspellchecker
````

````python
from spellchecker import SpellChecker

spell = SpellChecker()
````

````python
# Find those words that may be misspelled
misspelled = spell.unknown(['somtime', 'things', 'This', 'csn'])
misspelled
````

````python
# Find those words that have correct spelling
spell.known(['somtime', 'things', 'This', 'csn'])
````

````python
# Get the one `most likely` answer
for word in misspelled:
    print(spell.correction(word))
````

````python
# Get a list of `likely` options
for word in misspelled:
    print(spell.candidates(word))
````

## Named Entity Recognition with NLTK library


### Extract Entity Name and Entity Type
We can also extract the `label` of each Named Entity in corpus. For this, we will build out a data frame of all the Named Entities and their types.

````python
corpus = """In ancient Rome, some neighbors live in three adjacent houses.
In the center is the house of Senex, who lives there with wife Domina, son Hero, and several slaves, including head slave Hysterium and the musical's main character Pseudolus.
A slave belonging to Hero, Pseudolus wishes to buy, win, or steal his freedom. One of the neighboring houses is owned by Marcus Lycus, who is a buyer and seller of beautiful women; the other belongs to the ancient Erronius, who is abroad searching for his long-lost children (stolen in infancy by pirates). 
One day, Senex and Domina go on a trip and leave Pseudolus in charge of Hero. Hero confides in Pseudolus that he is in love with the lovely Philia, one of the courtesans in the House of Lycus (albeit still a virgin).
"""
````

````python
named_entities = []
for sentence in nltk.sent_tokenize(corpus):     # corpus includes all sentences as whole text
    sent_tokens = nltk.word_tokenize(sentence)  # tokenize each sent
    word_tag = nltk.pos_tag(sent_tokens)        # pos tag for all words of a sent
    for chunk in nltk.ne_chunk(word_tag):       # for each word (chunk) after name entity tag
        if hasattr(chunk, 'label'):             # it checks does chunk object have label attribute
            word = ' '.join(c[0] for c in chunk)
            named_entity_tag = chunk.label()
            temp_named_entity = (word, named_entity_tag)
            named_entities.append(temp_named_entity)
            temp_named_entity = ''              # or  temp_named_entity = None
            
df_entity_nltk = pd.DataFrame(named_entities, 
                            columns=['Entity Name', 'Entity Type'])
df_entity_nltk
````

### The most Frequent Named Entities

Let’s now find out **the most frequent Named Entities** in the corpus! For this, we can aggregate the data frame to find the top occuring entities and types.  


````python
# Get the top 5 Named Entities and types
top_entities = (df_entity_nltk.groupby(by=['Entity Name', 'Entity Type'])
                           .size()
                           .sort_values(ascending=False)
                           .reset_index().rename(columns={0 : 'Frequency'}))

top_entities.iloc[:5,:]   # or   top_entities.T.iloc[:,:5]
````

We can also group by the entity types to get a sense of what types of entites occur most in the corpus.


````python
# Get the top 5 Named Entity types
top_entities = (df_entity_nltk.groupby(by=['Entity Type'])
                           .size()
                           .sort_values(ascending=False)
                           .reset_index().rename(columns={0 : 'Frequency'}))
top_entities.iloc[:5,:]     
````

We can see that people, places and organizations are the most mentioned entities though.


## Named Entity Recognition with spaCy library 

spaCy is a free, open-source library for advanced NLP task in Python. It has some excellent capabilities for named entity recognition. spaCy models are trained to support the following entity types:

<img src="../images/spaCy Named Entity Recognition.png" style="height: 450px;">

We shall work with the `en_core_web_sm` model of spaCy. `en_core_web_sm` is an English multi-task [convolutional neural network (CNN)](https://en.wikipedia.org/wiki/Convolutional_neural_network), a type of machine learning model, trained on OntoNotes. The model can assign context-specific token vectors, POS tags, dependency parse and named entities to text. We can download the model through spaCy, and load it using `en_core_web_sm.load()`.
 
Let us try and use spaCy named entity recognition.


````python
# Uncomment and run the following code to install spacy and download the model
# !pip install spacy
# !python -m spacy download en_core_web_sm
````

````python
import spacy
import en_core_web_sm

# Load English tokenizer, tagger, parser, NER and word vectors
nlp = en_core_web_sm.load()
````

After loading the spaCy model in the variable `nlp`, we can give a text as input to the model. The model will process the given text, in which we can extract the entities present in the text. The property `.ent_type_` returns the entity type of the given word.


````python
# sample_text="Sara likes to visit different parts of malaysia. She visits Cameron hilands, Frozen hills, Port dickson and Melaka until now."
# sample_text=" I'm going to Germany this Monday."
# sample_text="Mark and John are working at Google."
# sample_text="WASHINGTON -- In the wake of a string of abuses by New York police officers in the 1990s, Loretta E. Lynch, the top federal prosecutor in Brooklyn, spoke forcefully about the pain of a broken trust that African-Americans felt and said the responsibility for repairing generations of miscommunication and mistrust fell to law enforcement."
sample_text = "When, after the 2010 election, Wilkie, Rob, Oakeshott, Tony Windsor and the Greens agreed to support Labor, they gave just two guarantees: confidence and supply."

doc = nlp(sample_text)
print(doc)

# print named entities 
print([(word, word.ent_type_) for word in doc if word.ent_type_])
````

### Extract Entity Name and Entity Type

Here we will learn how to extract the name and type of entites in a corpora. We can do this by iterating through each sentence in the corpora, feeding the sentences into the spaCy model to find the entities in the sentences. As the model returns as output the word and entity tag of each word, we use the `.text` attribute to get the actual word and the `.ent_type_` attribute to get the entity type.

We can use a data frame to display all the entities that we have extracted.


````python
# Extract the `label` of each Named Entity with spaCy

corpora = ["In ancient Rome, some neighbors live in three adjacent houses.",
"In the center is the house of Senex, who lives there with wife Domina, son Hero, and several slaves, including head slave Hysterium and the musical's main character Pseudolus.",
"A slave belonging to Hero, Pseudolus wishes to buy, win, or steal his freedom. One of the neighboring houses is owned by Marcus Lycus, who is a buyer and seller of beautiful women; the other belongs to the ancient Erronius, who is abroad searching for his long-lost children (stolen in infancy by pirates). ",
"One day, Senex and Domina go on a trip and leave Pseudolus in charge of Hero. Hero confides in Pseudolus that he is in love with the lovely Philia, one of the courtesans in the House of Lycus (albeit still a virgin)."]

named_entities = []
for sentence in corpora:     # corpora includes a list of sentences
    temp_named_entity = ''   # or  temp_named_entity = None
    sentence = nlp(sentence)
    for word in sentence:
        term = word.text           # each word
        tag = word.ent_type_       # name entity tag
        if tag:
            temp_named_entity = (term, tag)
            named_entities.append(temp_named_entity)
            temp_named_entity = ''
            
df_entity = pd.DataFrame(named_entities, 
                            columns=['Entity Name', 'Entity Type'])
df_entity            
````
