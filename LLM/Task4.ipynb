Cell 2 — Load the pretrained model

```python

from transformers import pipeline

model_name = "distilbert/distilbert-base-uncased-finetuned-sst-2-english"

classifier = pipeline(
    task="sentiment-analysis",
    model=model_name
)

print("Pretrained sentiment model loaded successfully!")
```

Cell 3 — Test sample customer reviews

```python

reviews = [
    "This product is excellent. I really love it!",
    "The quality is terrible and I regret buying it.",
    "The delivery was fast, but the product was disappointing.",
    "I expected better quality for this price.",
    "Absolutely fantastic service. Highly recommended!"
]

results = classifier(reviews)

for review, result in zip(reviews, results):
    print("Review:", review)
    print("Prediction:", result["label"])
    print("Model score:", round(result["score"] * 100, 2), "%")
    print("-" * 60)
```

The model receives text, processes it through a tokenizer and Transformer network, and returns a sentiment label and score.

Cell 4 — Let students enter their own text

```python

while True:
    text = input("Enter a review (or type 'exit'): ")

    if text.lower().strip() == "exit":
        break

    if not text.strip():
        print("Please enter some text.")
        continue

    result = classifier(text)[0]

    print("\nPredicted sentiment:", result["label"])
    print("Model score:", round(result["score"] * 100, 2), "%")
    print()
```

3. Advanced mini-experiment: Compare two models
This extension connects the practical exercise to the comparative analysis part of your training outline.

```python

from transformers import pipeline

bert_model = pipeline(
    "sentiment-analysis",
    model="distilbert/distilbert-base-uncased-finetuned-sst-2-english"
)

multilingual_model = pipeline(
    "sentiment-analysis",
    model="nlptown/bert-base-multilingual-uncased-sentiment"
)

samples = [
    "I absolutely love this product.",
    "This is the worst experience ever.",
    "Me gusta mucho este producto."
]

for sentence in samples:
    print("\nText:", sentence)

    print("DistilBERT:", bert_model(sentence)[0])
    print("Multilingual BERT:", multilingual_model(sentence)[0])
```

The second model returns a 1–5 star rating, rather than a positive/negative label. Its output therefore needs to be interpreted differently. The Spanish example also illustrates why language coverage matters: a model trained for English sentiment should not be assumed to work reliably in other languages

Cell 2 — Load the pretrained GPT-2 model

```python

import torch
from transformers import pipeline

model_name = "openai-community/gpt2"

generator = pipeline(
    "text-generation",
    model=model_name,
    device=-1
)

print("GPT-2 loaded successfully!")
print("Device: CPU")
```

Cell 3 — Generate your first AI text

```python

prompt = "Artificial intelligence is changing education because"

result = generator(
    prompt,
    max_new_tokens=60,
    do_sample=False,
    pad_token_id=generator.tokenizer.eos_token_id
)

print("INPUT PROMPT:")
print(prompt)

print("\nGENERATED TEXT:")
print(result[0]["generated_text"])
```

**What Should Students Observe?**
- The generated text begins with the input prompt.
- The model continues the text based on the given prompt.
- When the same prompt and settings are used with greedy decoding, the model generally produces the same output.
- The quality and relevance of the generated text depend on the prompt and the model's training.
- Changing the prompt or generation parameters can affect the output.
Note: GPT-2 is a text-completion model. It predicts and generates the continuation of a given text. Unlike modern instruction-tuned chat models, it is not specifically designed to answer every question directly or accurately.

Cell 4 — Testing Different prompts
**Explanation:** In this experiment, students will observe how changing the input prompt affects the text generated by the model.

```python

prompts = [
    "The future of technology",
    "Once upon a time, in a small village",
    "Students can improve their learning by",
    "Climate change affects the world because"
]

for prompt in prompts:
    result = generator(
        prompt,
        max_new_tokens=40,
        do_sample=False,
        pad_token_id=generator.tokenizer.eos_token_id
    )

    print("\nPROMPT:", prompt)
    print("OUTPUT:", result[0]["generated_text"])
    print("-" * 70)
```

Cell 5 — comparing effect of temperature
**Explanation:** In this experiment, students will observe how temperature affects the text generated by the model. Lower temperatures generally produce more predictable and consistent outputs, while higher temperatures can increase variety and randomness in the generated text.

```python

prompt = "A successful student should"

for temperature in [0.3, 0.7, 1.0]:
    result = generator(
        prompt,
        max_new_tokens=45,
        do_sample=True,
        temperature=temperature,
        pad_token_id=generator.tokenizer.eos_token_id
    )

    print(f"\nTemperature: {temperature}")
    print(result[0]["generated_text"])
    print("-" * 60)
```

Ask students to compare the outputs. A higher temperature does not necessarily mean better quality; it can also make the generated text more unpredictable.

Cell 6 — Interactive text generator
This cell makes the exercise interactive, allowing students to test their own ideas, story openings, and educational topics.

```python

while True:
    prompt = input(
        "\nEnter a prompt (or type 'exit' to stop): "
    ).strip()

    if prompt.lower() == "exit":
        print("Exercise completed.")
        break

    if not prompt:
        print("Please enter a prompt.")
        continue

    result = generator(
        prompt,
        max_new_tokens=60,
        do_sample=True,
        temperature=0.7,
        pad_token_id=generator.tokenizer.eos_token_id
    )

    print("\nGenerated text:")
    print(result[0]["generated_text"])
```

### Important Parameters in GPT-2 Text Generation

**1. `model_name`**

Specifies the name or identifier of the pretrained language model to load. In this exercise, we use GPT-2.

**2. `max_new_tokens=60`**

Sets the maximum number of new tokens the model can generate after the input prompt. A token may be a word, part of a word, or a punctuation mark.

**3. `do_sample=False`**

Disables random sampling. The model uses greedy decoding, selecting the highest-probability next token at each generation step.

**4. `do_sample=True`**

Enables random sampling from the model's next-token probability distribution. This can produce different outputs for the same prompt.

**5. `temperature=0.7`**

Controls the randomness of token selection when sampling is enabled. Lower values generally produce more predictable outputs, while higher values can increase variety and unpredictability. Temperature has no effect when `do_sample=False`.

**6. `pad_token_id`**

Specifies the token ID used for padding. In this exercise, GPT-2's end-of-sequence token ID is used to handle padding-related settings during text generation.

### How Does GPT-2 Generate Text?

GPT-2 is an autoregressive language model. It generates text by predicting the next token based on the preceding tokens. It repeatedly adds new tokens until it reaches the specified generation limit or another stopping condition.

### Key Takeaway

Changing generation parameters can affect the length, predictability, and variety of generated text. However, more varied output does not necessarily mean better-quality text.
