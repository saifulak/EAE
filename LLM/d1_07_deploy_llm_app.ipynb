## 4. Deployment concepts
- `st.text_area` collects input and `st.button` triggers the action.
- `@st.cache_resource` prevents loading a new model on every rerun.
- The app runs on your computer by default; that is local deployment, not public hosting.
- Do not paste private customer data into a public demo. Add authentication and appropriate data controls before using real business data.
