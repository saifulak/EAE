# Introduction to NLP
NLP is a branch of data science that consists of systematic processes for analyzing, understanding, and deriving information from the text data in a smart and efficient manner. By utilizing NLP and its components, one can organize the massive chunks of text data, perform numerous automated tasks and solve a wide range of problems such as – automatic summarization, machine translation, named entity recognition, relationship extraction, sentiment analysis, speech recognition, and topic segmentation etc.


## What is NLP?
NLP is a discipline that focuses on the interaction between data science and human (natural) language. It involves the interaction between humans and computers using natural languages. Most NLP uses machine learning techniques to process and analyze the large amounts of text data. The objective of NLP is to give computers the ability to read, understand, interpret and derive meaning from human languages. Today, NLP is blooming thanks to the improvements in computational power and the vast amount of data available, and is scaling to lots of industries.


## NLP Applications and Tasks
The value of NLP comes from the use cases, usually involves the automatic handling of natural human language like text or speech, to help with lots of tasks ona daily basis. Examples of NLP applications:
- E-mail applications such as Gmail and Outlook use NLP to determine whether an email is spam or valid
- Language translation applications such as Google Translate use NLP to translate one language to another
- Word processors and text editors such as Microsoft Word and Grammarly use NLP to check all types of mistakes in text, from typing errors to grammatical errors
- Personal assistant applications such as Siri and Alexa use NLP for speech recognition and question answering
- Chatbots used on several channels, such as the Internet, applications, and messaging platforms are created using NLP

The general tasks that can be solved with NLP are:
- Machine translation
- Speech-to-text and text-to-speech conversion
- Sentiment analysis
- Word Prediction
- Topic discovery and modeling
- Grammatical Error Correction
- Automatic summarization
- Voice Recognition

## Top 10 Python NLP Libraries 

### 1) Natural Language Toolkit (NLTK) module with Python

The NLTK module is a massive tool kit, aimed at helping you with the entire Natural Language Processing (NLP) methodology. It is an open source library in python. NLTK will aid you with everything from splitting sentences from paragraphs, splitting up words, recognizing the part of speech of those words, highlighting the main subjects, and then even with helping your machine to understand what the text is all about
NLTK is a leading platform for building Python programs to work with human language data. It provides easy-to-use interfaces to over 50 corpora and lexical resources such as WordNet, along with a suite of text processing libraries for classification, tokenization, stemming, etc. This library provides a practical introduction to programming for language processing. NLTK has been called “a wonderful tool for teaching and working in computational linguistics using Python,” and “an amazing library to play with natural language.”


#### Advantages of NLTK
- Has support for most NLP tasks
- Provide acsses to numerous text corpora

### 2) Gensim
Gensim is a Python library for topic modelling, document indexing and similarity retrieval with large corpora. Target audience is basically the natural language processing (NLP) and information retrieval (IR) community. The features of this library include such as all algorithms are memory-independent w.r.t. the corpus size, intuitive interfaces, efficient multicore implementations of popular algorithms, distributed computing, etc.

### 3) polyglot
Polyglot is a natural language pipeline which supports massive multilingual applications. The features include tokenisation, language detection, named entity recognition, part of speech tagging, sentiment analysis, word embeddings, etc.

It provides a simple API for diving into common natural language processing (NLP) tasks such as part-of-speech tagging, noun phrase extraction, sentiment analysis, classification, translation, WordNet integration, parsing, word inflection, adds new models or languages through extensions, and more.

### 5) CoreNLP
Stanford CoreNLP provides a set of human language technology tools. Stanford CoreNLP’s goal is to make it very easy to apply a bunch of linguistic analysis tools to a piece of text. Stanford CoreNLP integrates many of Stanford’s NLP tools, including the part-of-speech (POS) tagger, the named entity recognizer (NER), the parser, the coreference resolution system, sentiment analysis, bootstrapped pattern learning, and the open information extraction tools. The tools variously use rule-based, probabilistic machine learning, and deep learning components.

### 6) spaCy
spaCy is a library for advanced Natural Language Processing in Python and Cython which comes with a number of interesting features. spaCy comes with pre-trained statistical models and word vectors, and currently supports tokenization for 49+ languages. It features state-of-the-art speed, convolutional neural network models for tagging, parsing and named entity recognition and easy deep learning integration.

### 7) Pattern
Pattern is a web mining module for the Python programming language. It has tools for data mining (Google, Twitter, and Wikipedia API, a web crawler, an HTML DOM parser), natural language processing (part-of-speech taggers, n-gram search, sentiment analysis, WordNet), machine learning (vector space model, clustering, SVM), network analysis by graph centrality and visualization. Pattern supports Python 2.7 and Python 3.6.

### 8) Vocabulary
Vocabulary is a Python library for natural language processing which is basically a dictionary in the form of Python module. Using this library, for a given word you can get its meaning, synonyms, antonyms, part of speech, translations and other such. This library is easy to install and is a decent substitute to Wordnet.

### 9) PyNLPl
PyNLPl, pronounced as ‘pineapple’, is a Python library for Natural Language Processing. It contains various modules useful for common, and less common, NLP tasks. PyNLPl can be used for basic tasks such as the extraction of n-grams and frequency lists, and to build a simple language model. PyNLPl most notably features a very extensive library for working with FoLiA XML (Format for Linguistic Annotation).

### 10) Quepy
Quepy is a python framework to transform natural language questions into queries in a database query language. It can be easily customized to different kinds of questions in natural language and database queries. Quepy uses an abstract semantics as a language-independent representation that is then mapped to a query language. This allows your questions to be mapped to different query languages in a transparent manner.

# Feature Extraction

Feature extraction occurs in the context of machine learning applications. It means to **extract** and **produce** feature representations that are appropriate for the type of NLP task from a text.

Let to explain some terms:
- Tokenization – process of converting a text into *tokens*
- Tokens – words or sentences or entities present in the text
- Lexicon – Words and their meanings. Example: English dictionary. Consider, however, that various fields will have different lexicons.
- Corpus – Body of text, singular. Corpora is the plural of this. Example: A collection of medical journals.

## Tokenization
A word (token) is the minimal unit that a machine can understand and process. Therefore, any text string cannot be further processed without going through tokenization. Tokenization is the process of splitting the given text into smaller pieces called *tokens*. Words, numbers, punctuation marks, and others can be considered as tokens.

Let's show an example of how one might actually tokenize something into tokens with the NLTK module.

```python
from nltk.tokenize import sent_tokenize, word_tokenize

# example_text = "Paragraphs can contain many different kinds of information. A paragraph could contain a series of brief examples or a single long illustration of a general point. It might describe a place, character, or process; narrate a series of events; compare or contrast two or more things; classify items into categories; or describe causes and effects. Regardless of the kind of information they contain, all paragraphs share certain characteristics. One of the most important of these is a topic sentence."
example_text = "We are attacking on their left flank but are losing many men. We cannot see the enemy army. Nothing else to report. We are ready to attack but are waiting for your orders."
# example_text = "Here are some very simple basic sentences. They won't be very interesting, I'm afraid.", The point of these examples is to _learn how basic text cleaning works_ on *very simple* data."

sen_tokens = sent_tokenize(example_text)
print(sen_tokens)

word_tokens = word_tokenize(example_text)
print(word_tokens)
```

```python
raw_docs = ["Here are some very simple basic sentences.",
"They won't be very interesting, I'm afraid.",
"The point of these examples is to _learn how basic text cleaning works_ on *very simple* data."]

docs_tokens = [word_tokenize(doc) for doc in raw_docs]
print(docs_tokens)
```

## Phrase Matching
Phrase Matching is to define some patterns and then see which phrases matched with the patterns we defined.

But before starting the Phrase Matching concept, we need to be familiar with Regular Expression.

### Regular Expression

Regular expression is a sequence of character(s) mainly used to find and replace patterns in a string or file. They are supported by most of the programming languages like Python, Perl, R, Java and many others.

Regular expressions (<a href="https://docs.python.org/3/library/re.html" target="_blank" rel="noopener nofollow">Regular expressions in Python 3</a>) use two types of characters:

- <font color=green><b>a) Meta characters</b></font>: these characters have a special meaning. Here’s a complete list of them:
<font color=red><b>. ^ $ * + ? { } [ ] | ( ) \ </b></font>
    - <b>Character matches:</b>
    
        <font color=red><b>.</b></font> :       Matches with any single character except newline
        
        <font color=red><b>^</b></font> :       Match the start of the string
        
       <font color=red><b>$</b></font> :       Match the end of the string
      
       <font color=red><b>[ ]</b></font> :    Matches any single character in a square bracket
      
      <b>[a-z]</b> :    Matches one of the range of character a,b,...,z
      
      <b>[^abc]</b> : Matches a character that is not in a,b or c
      
      <b>a<font color=red>|</font>b</b> :     Matches either a or b, where a and b are string
      
      <font color=red><b>( )</b></font> :     Groups regular expressions and returns matched text
      
      <font color=red><b>\\</b></font> :       It is used for special meaning characters
    <br><br>
    - <b> Characters symbols</b>
    
        <font color=red><b> \b </b></font>: Matches word boundary
        
        <font color=red><b> \d </b></font>: Any digit, equivalent to [0-9]
        
        <font color=red><b> \D </b></font>: Any non-digit, equivalent to [^ 0-9]
        
        <font color=red><b> \s </b></font>: Any whitespace, equivalent to [\t\n\r\f\v]
        
        <font color=red><b> \S </b></font>: Any non-whitespace, equivalent to [^ \t\n\r\f\v]
        
        <font color=red><b> \w </b></font>: Alphanumeric character, equivalent to [a-zA-z0-9_ ]
        
        <font color=red><b> \W </b></font>: Non-alphanumeric character, equivalent to [^ a-zA-z0-9_ ]
<br><br>
    - <b> Repetitions:</b>

        <font color=red><b>*</b></font> :       0 or more occurrences
        
        <font color=red><b>+</b></font> :       1 or more occurrences 
          
        <font color=red><b>?</b></font> :       0 or 1 occurrence
        
        <font color=red><b>{</font><b>n<font color=red>}</b></font> :       Exactly n repetitions, n>=0
        
        <font color=red><b>{</font><b>n, <font color=red>}</b></font> :       At least n repetitions
        
        <font color=red><b>{</font><b> ,n<font color=red>}</b></font> :       At most n repetitions
    

- <font color=green><b>b) Literals </b></font>(like a,b,1,2…)

In Python, we have module `re` that helps with regular expressions. So you need to import library re before you can use regular expressions in Python.

```python
import re
```

### The most common uses of regular expressions
- <font color=blue>Search a string </font>(search and match)
- <font color=blue>Finding a string </font>(findall)
- <font color=blue>Break string into a sub strings</font> (split)
- <font color=blue>Replace part of a string </font>(sub)

The most commonly used methods which The ‘re’ package provides to perform queries on an input string:

- <font color=green><b>re.match(pattern, string)</b></font>: 
This method finds match if it occurs at start of the string.

- <font color=green><b>re.search(pattern, string)</b></font>: 
This method is similar to *match()* but it doesn’t restrict us to find matches at the beginning of the string only. The *search()* method is able to find a pattern from any position of the string but it only returns the first occurrence of the search pattern.

- <font color=green><b>re.findall (pattern, string)</b></font>: 
This method helps to get a list of all matching patterns. It has no constraints of searching from start or end.

- <font color=green><b>re.split(pattern, string, maxsplit=0)</b></font>: 
This methods helps to split string by the occurrences of given pattern.

- <font color=green><b>re.sub(pattern, replace, string)</b></font>:
This method helps to search a pattern and replace with a new sub string. If the pattern is not found, string is returned unchanged.

```python
# match()
result = re.match(r'Te', 'Text Analytic and Text Mining')
result
```

```python
# match()
value = "vrheesville"
m = re.match(r"voo", value)
m
```

```python
# match()
value = "vrheesville"
m = re.match(r"vo+", value)
m
```

```python
# search()
result = re.search(r'Te', 'Text Analytic and Text Mining')
result
```

```python
# findall()
result = re.findall(r'Te', 'Text Analytic and Text Mining')
result
```

```python
# findall()
result = re.findall(r'(Te.*)', 'Text Analytic and Text Mining')
result
```

## Number of Words & Sentences

We are interested in finding the number of words or number of sentences in a text. We can do this by tokenizing the text, and then count the number of tokens.

```python
example_text = "We are attacking on their left flank but are losing many men. We cannot see the enemy army. Nothing else to report. We are ready to attack but are waiting for your orders."

sen_tokens = sent_tokenize(example_text)
word_tokens = word_tokenize(example_text)
```

```python
print("Number of sentences:\t", len(sen_tokens))  # Number of Sentences
print("Number of words:\t", len(word_tokens))       # Number of Words
```

To obtain the number of times a specific word occurs across the whole text (total number of occurrences), we can use the `count()` function.

```python
print("Number of occurrences of \'{}\':\t{}".format('the', word_tokens.count('the')))
print("Number of occurrences of \'{}\':\t{}".format('We', word_tokens.count('We')))
print("Number of occurrences of \'{}\':\t{}".format('not', word_tokens.count('not')))
```

## Stop-words
For finding stop-words in text, we first need to know the meaning of stop-words.

### Language Stop-words
Stop-words usually refer to the most common words in a language. For instance, the English language contains stop words like ‘a’, ‘an’, ‘are’, ‘as’, ‘at’, ‘be’, ‘by’, ‘for’, ‘from’, ‘has’, ‘he’, ‘is’, ‘in’, ‘it’, ‘its’, ‘of’, ‘on’, ‘that’, ‘the’, ‘to’, ‘was’, ‘were’, ‘will’, ‘with’, etc. 

The NLTK module has a list of stop-words stored in 16 different languages. We can import the `stopwords` module to get the list of stop-words defined in the module.

```python
# nltk.download('stopwords')
from nltk.corpus import stopwords

stop_words_list = set(stopwords.words('english'))
print(stop_words_list)
print("Number of stop-words:\t", len(stop_words_list))
```

Let's count the number of words in the text that are in the stop-words list.

```python
example_text = "We are attacking on their left flank but are losing many men. We cannot see the enemy army. Nothing else to report. We are ready to attack but are waiting for your orders."
word_tokens = word_tokenize(example_text)

# Number of stop-words 
stop_word_accumulator = 0
for word in word_tokens:
    if word in stop_words_list:
        stop_word_accumulator += 1

print("Number of stop-words in text:\t", stop_word_accumulator)
```

## Extract the Numbers

Sometimes, we want to extract the numbers containing in a text. We can either do this with the `isdigit()` function which checks if the given variable is a digit, or with the `re.findall()` function by specifiying the search parameter to just digits.

```python
text = "hi 6543 23 cat 25.5 rabbit 31 6 dog' * 450"
# text = 'he33llo 42 I\'m a 32 string 30'
```

```python
res = [int(word) for word in word_tokenize(text) if word.isdigit()]
print("The numbers list is :\t", res)
```

### Long Words
Long words are words that are more than 3 letters long.

```python
text1 = "We are attacking on their left flank but are losing many men. We cannot see the enemy army. Nothing else to report. We are ready to attack but are waiting for your orders."
text2 = word_tokenize(text1)
```

### Capitalized Words
Capitalized words are the words that start by capital (uppercase) letter A-Z.

There are several built-in methods that we can use on strings to identify if the words are capitalized:
- <font color=green><b>str.istitle()</b></font>: 
This method returns **True** if the string is a capitalized, **False** if the string is not capitalized or is empty.

- <font color=green><b>str.isupper()</b></font>:
This method returns **True** if all characters in the string are uppercase characters, **False** if any characters in a string are lowercase character.

- <font color=green><b>str.islower()</b></font>:
This method returns **True** if all characters in the string are lowercase characters, **False** if any characters in a string are uppercase character.

```python
text1 = "We are attacking on their left flank but are losing many men. We cannot see the enemy army. Nothing else to report. We are ready to attack but are waiting for your orders."
text2 = word_tokenize(text1)

# Capitalized words
title = [w for w in text2 if w.istitle()]      
upper = [w for w in text2 if w.isupper()]
lower = [w for w in text2 if w.islower()]

print("Capitalized words:\t", title)
print("Uppercase words:\t", upper)
print("Lowercase words:\t", lower)
```

There are also built-in methods that we can use to modify strings, turning them into capitalized, uppercased or lowercased words:
- <font color=green><b>str.title()</b></font>:
This method returns the capitalized version of the string.

- <font color=green><b>str.upper()</b></font>:
This method returns the uppercased string from the given strin (converts all lowercase characters to uppercase).

- <font color=green><b>str.lower()</b></font>:
This method returns the lowercased string from the given string (converts all uppercase characters to lowercase).

```python
text_upper = text1.upper()
text_upper
```

```python
text_upper.lower()
```

## Non ASCII Characters

Some text data that we scrape from the web has symbols such as ™ and £. These characters are called **non-ASCII characters**.

```python
import re

text = "Some™ sentence£ from® dirty⅙ dataset© has♫ non¥ ascii÷ characters¤"
print(text)
```

### Extract non ASCII Characters

In some cases, non-ASCII characters are useful such as getting the currency type or mathematical symbols from the text. ASCII characters have their unicode within the range of *x00-x7f*. So to extract the non-ascii characters, we can use the `re.findall()` function to find all the characters which are **not** within that range.

```python
nonascii = re.findall(r'[^\x00-\x7f]', text)
print(nonascii)
```

# Frequency Distributions of Words
A frequency distribution records the number of times each word has occurred in a text. A frequency distribution can be defined as a function mapping from each word to the number of times that word occurred as an outcome.

The `FreqDist` class is used to encode the frequency distributions. It works like a dictionary, where the keys are the words in the text and the values are the frequency count associated with that word.

- <font color=green><b>FreqDist(text)</b></font>: 
This method returns the frequency distribution of the text as a dictionary.

```python
from nltk.probability import FreqDist
```

We want to find the frequency distributions of words for following text.

    text = "The local part of an email address has no significance for intermediate mail relay systems other than the final mailbox host. Email senders and intermediate relay systems must not assume it to be case-insensitive, since the final mailbox host may or may not treat it as such. A single mailbox may receive mail for multiple email addresses, if configured by the administrator. Conversely, a single email address may be the alias to a distribution list to many mailboxes"

```python
text = "The local part of an email address has no significance for intermediate mail relay systems other than the final mailbox host. Email senders and intermediate relay systems must not assume it to be case-insensitive, since the final mailbox host may or may not treat it as such. A single mailbox may receive mail for multiple email addresses, if configured by the administrator. Conversely, a single email address may be the alias to a distribution list to many mailboxes"
text_tokens = word_tokenize(text)
```

```python
text_tokens1 = [word.lower() for word in text_tokens]

dist = FreqDist(text_tokens1)

print("Number of unique tokens in the text:\t", len(dist))
dist
```

```python
# Finding frequency distribution in one line
dist = FreqDist(word.lower() for word in word_tokenize(text))
```

```python
print("Number of tokens in the text:\t", len(text_tokens1))
```

```python
# frequency distribution of a word
print("Frequency distribution of \'{}\':\t{}".format('the', dist['the']))
print("Frequency distribution of \'{}\':\t{}".format('not', dist['not']))
print("Frequency distribution of \'{}\':\t{}".format('mail', dist['mail']))
```

## FreqDist Methods
Here are the most commonly used methods to manipulate or extract information from the `FreqDist` class:

- keys()
- values()
- N()
- B()
- hapaxes()
- Nr()
- freq()
- max()
- pprint()
- most_common()
- plot()

To have a copy of the `FreqDist` list of keys (words), we can use the `keys()` method.

```python
vocab = dist.keys()
vocab
```

The `values()` method returns a copy of the `FreqDist` list of values which is the frequency distributions of each word.

```python
dist.values()
```

The `N()` method returns the total number of word outcomes that have been recorded by this `FreqDist`.

```python
dist.N()
```

For the number of unique words with counts greater than zero, we use the `B()` method. It is the same as `len(FreqDist)`.

```python
dist.B()
```

The `hapaxes()` method returns a list of all words that occur only once.

```python
print(dist.hapaxes())
```

To get the number of words with a specific frequency, we use the `Nr()` method.

```python
dist.Nr(3) 
```

The `freq()` method returns the frequency of a given word. 

The frequency of a word is defined as the count of that word (the number of times that word occur) divided by the total number of words that have been recorded by this `FreqDist`.

```python
dist.freq('the')
```

To find the word with the greatest number of outcome in this frequency distribution, we can use the `max()` method. If two or more words have the same number of outcome, it returns only one of them.

```python
dist.max()  
```

The `pprint()` method takes an integer value for the *maxlen* parameter and returns a maximum of *maxlen* number of items in the same form of `FreqDist`.

```python
dist.pprint(maxlen=10)
```

To obtain the list of the *n* most common words and their counts (from the most common to the least), we can use the `most_common(n)` method. If *n* is not specified, `most_common()` returns all words in the counter.

```python
# List of the n most common words
n = 7
print(dist.most_common(n))
```

```python
# List of the n least common words
n = 4
print(dist.most_common()[:-n-1:-1])
```

The `plot()` method plots words from the frequency distribution, displaying the most frequent word first. 

```python
%matplotlib inline
dist.plot()
```

If an integer parameter is supplied, the method will stop after this many words have been plotted.

```python
dist.plot(10, title = '10 most frequently occurred word')
```
